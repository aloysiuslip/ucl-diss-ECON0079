```python
# Write Python code to list file structure within '1_FAME_raw_data'
# If there are a large number of files in a folder following a similar structure, truncate

import os

# Count the number of files in the current directory,
# and if there are more than 10 files, truncate the output to show only the first 10 files.
# and write instead '[... 2, 3, 4]' to indicate that there are more files which vary in just the suffix or whichever.
# Do this as well with the folder name if that's similar, we don't need to list the variants of subfiles for these.
# This is just to give an AI a sense of the file structure, not to list every single file.
# Don't add a newline between every subdirectory. Don't be exhaustive.
def list_file_structure(directory) -> None:
    for root, dirs, files in os.walk(directory):
        # Print the current directory
        print(f"Directory: {root}")
        
        # Print the subdirectories
        if dirs:
            print("Subdirectories:")
            for d in dirs[:10]:  # Limit to first 10 subdirectories
                print(f"  {d}")
            if len(dirs) > 10:
                print(f"  [... {len(dirs) - 10} more subdirectories]")
        
        # Print the files
        if files:
            print("Files:")
            for f in files[:10]:  # Limit to first 10 files
                print(f"  {f}")
            if len(files) > 10:
                print(f"  [... {len(files) - 10} more files]")
        
        print()  # Add a newline for better readability

list_file_structure('1_FAME_raw_data')
```
